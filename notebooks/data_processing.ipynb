```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
```

```python
import os
print(os.getcwd())
```

```python
df = pd.read_csv("Churn_Modelling.csv")
df.head()
```

```python
df = df.drop(["RowNumber", "CustomerId", "Surname"], axis = 1)
df.head()
```

```python
# Encoding categorical variables
label_encode = LabelEncoder()

df["Gender"] = label_encode.fit_transform(df["Gender"])
```

```python
# Onehot encoding for categorical variable : Geography (more than 2 classes)
onehot_encode = OneHotEncoder(sparse_output = False)

geo_encoder = onehot_encode.fit_transform(df[["Geography"]])
geo_encoder
```

```python
onehot_encode.get_feature_names_out(["Geography"])
```

```python
geo_encoder_df = pd.DataFrame(geo_encoder, columns = onehot_encode.get_feature_names_out(["Geography"]))
geo_encoder_df
```

```python
df = pd.concat([df.drop(["Geography"], axis = 1), geo_encoder_df], axis = 1)
df.head()
```

```python
import pickle

with open("label_encode.pkl", 'wb') as file:
    pickle.dump(label_encode, file)

with open("onehot_encode.pkl", 'wb') as file:
    pickle.dump(onehot_encode, file)
```

```python
# Splitting the dataset into input and output features
x = df.drop("Exited", axis = 1)
y = df["Exited"]

# Splitting the data into training and testing dataset
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size =  0.2, random_state = 42)

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
with open("scaler.pkl", 'wb') as file:
    pickle.dump(scaler, file)
```
