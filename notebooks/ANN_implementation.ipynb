```python
import pandas as pd
import numpy as np
import pickle

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime
```

```python
x_train = np.load(r"C:\Users\dhanu\OneDrive\Desktop\Udemy Course\Section_13\ANN_project\data\split data\x_train.npy")
x_test = np.load(r"C:\Users\dhanu\OneDrive\Desktop\Udemy Course\Section_13\ANN_project\data\split data\x_test.npy")
y_train = np.load(r"C:\Users\dhanu\OneDrive\Desktop\Udemy Course\Section_13\ANN_project\data\split data\y_train.npy")
y_test = np.load(r"C:\Users\dhanu\OneDrive\Desktop\Udemy Course\Section_13\ANN_project\data\split data\y_test.npy")
```

```python
model = Sequential([
    Dense(64, activation = 'relu', input_shape = (x_train.shape[1],)),  # Input Layer -> Hidden Layer 1
    Dense(32, activation = 'relu'), # Hidden Layer 1 -> Hidden Layer 2
    Dense(1, activation = 'sigmoid') # Hidden Layer ->  Output Layer
])
```

```python
model.summary()
```

```python
optim = tf.keras.optimizers.Adam(learning_rate = 0.01)
```

```python
# Compiling the ANN

model.compile(optimizer = optim, loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
# Tensorboard

log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tf_callback = TensorBoard(log_dir = log_dir, histogram_freq = 1)
```

```python
early_stop_callback = EarlyStopping(monitor = 'val_loss', patience = 5, restore_best_weights = True)
```

```python
# Training the model

history = model.fit(
    x_train, y_train, validation_data = (x_test, y_test), epochs = 20, callbacks = [tf_callback, early_stop_callback]
)
```

```python
model.save("ANN.h5")
```
